```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file = pd.read_csv("istherecorrelation.csv", delimiter=";")
display(file)
plt.figure(figsize=(10,6))
scatter = plt.scatter(file["WO [x1000]"], file["NL Beer consumption [x1000 hectoliter]"], c=file["Year"], cmap="viridis")
plt.ylabel(r"NL Beer Consumption ($\times 10^5$ litres)")
plt.xlabel(r"Students in academic university education ($\times 10^3$)")
plt.title("Annual Beer Consumption of Students in NL")
plt.colorbar(scatter, label="Years (2006-2018)")
plt.tight_layout()
plt.savefig("beer_consumption_students_nl.png", dpi=300)

```
